# Week 1 - Data Cleaning and Feature Engineering

This notebook runs the production cleaning pipeline for the Hotel Booking Demand dataset.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'src'))

from clean_data import load_dataset, clean_and_engineer, build_segment_summary, write_cleaning_report
from config import CLEANED_DATA_PATH, SEGMENTS_PATH, ensure_directories

ensure_directories()
raw_df = load_dataset()
cleaned_df, cleaning_report = clean_and_engineer(raw_df)
segment_summary = build_segment_summary(cleaned_df)

cleaned_df.to_csv(CLEANED_DATA_PATH, index=False)
segment_summary.to_csv(SEGMENTS_PATH, index=False)
write_cleaning_report(cleaning_report, cleaned_df)

cleaned_df.shape
```
