# Week 2 - Exploratory Data Analysis

This notebook runs the production EDA pipeline and then opens the generated cleaned data for ad hoc inspection.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT / 'src'))

from eda_analysis import run_eda
from config import CLEANED_DATA_PATH, VISUALIZATIONS_DIR, EDA_REPORT_PATH, BUSINESS_RECOMMENDATIONS_PATH

run_eda()
df = pd.read_csv(CLEANED_DATA_PATH)
df.shape
```

## Key Performance Snapshot

```python
pd.Series({
    'bookings': len(df),
    'cancellation_rate': df['is_canceled'].mean(),
    'avg_adr': df['adr'].mean(),
    'avg_lead_time': df['lead_time'].mean(),
    'avg_total_stay': df['total_stay'].mean(),
})
```

## Segment Performance

```python
df.groupby(['traveller_segment', 'lead_time_bucket', 'customer_value_segment']).agg(
    bookings=('is_canceled', 'size'),
    cancellation_rate=('is_canceled', 'mean'),
    avg_adr=('adr', 'mean'),
    avg_booking_value=('booking_value', 'mean')
).sort_values('bookings', ascending=False).head(20)
```
